```python
# 案例展示：0.1 + 0.2 是否等於 0.3？
a = 0.1
b = 0.2
c = a + b

print(f"a + b 的顯示結果：{c}")
print(f"a + b == 0.3 的判斷結果：{a + b == 0.3}")
```

# 可以什麼方式解決

```python
from decimal import Decimal

# 必須以字串型態傳入，避免在傳入前就變成 float 的近似值
a = Decimal('0.1')
b = Decimal('0.2')
print(a + b == Decimal('0.3'))  # 輸出：True
```

# Mission 4：算術運算與模擬成績
### 核心觀念：利用基礎加減乘法進行數值計算。

```python
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

base_score = 50
study_effect = 3 * study_hours             # TODO：3 × 讀書時數
sleep_effect = 2 * sleep_hours             # TODO：2 × 睡眠時數
screen_effect = -1.2 * screen_time_hours   # TODO：-1.2 × 螢幕使用時數
attendance_effect = 0.15 * attendance_rate  # TODO：0.15 × 出席率

sample_score = base_score
sample_score = sample_score + study_effect
sample_score = sample_score + sleep_effect
sample_score = sample_score + screen_effect
sample_score = sample_score + attendance_effect
print("模擬成績：", round(sample_score, 1))
```

# Mission 5：條件判斷與自訂函式
### 核心觀念：def 定義函式，return 回傳結果，Python 依靠縮排區隔邏輯區塊。

```python
def describe_score(score):
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

print("82分：", describe_score(82))
print("70分：", describe_score(70))
print("55分：", describe_score(55))
```

 # Mission 7：建立模擬的大學學習資料

```python

import numpy as np
import pandas as pd

# 建立隨機數產生器
rng = np.random.default_rng(42)

# 設定學生人數
student_count = 60
n = student_count

# 1. 產生各欄位資料
student_ids = np.arange(1, n + 1)

departments_list = rng.choice(
    ["資訊管理", "企業管理", "金融管理"],
    size=n
)

# 常態分布生成與邊界限制
sleep_data = np.round(
    rng.normal(loc=6.7, scale=1.1, size=n), 1
)

screen_data = np.round(
    np.clip(
        rng.normal(loc=5.0, scale=1.5, size=n),
        1.0, 10.0
    ), 1
)

study_data = np.round(
    np.clip(
        rng.normal(loc=2.5, scale=1.0, size=n),
        0.5, 6.0
    ), 1
)

attendance_data = np.round(
    np.clip(
        rng.normal(loc=88, scale=8, size=n),
        60, 100
    )
).astype(int)

# 2. 組合為 Pandas DataFrame
df = pd.DataFrame({
    "student_id": student_ids,
    "department": departments_list,
    "sleep_hours": sleep_data,
    "screen_time_hours": screen_data,
    "study_hours": study_data,
    "attendance_rate": attendance_data
})

# 顯示前五筆資料
df.head()
```

# Mission 8：建立小考成績欄位

```python
#w06-homework-1
import numpy as np
import pandas as pd

# 設定顯示列數，讓 60 筆資料能完整以 HTML 表格呈現而不被截斷
pd.set_option("display.max_rows", 60)

# 1. 初始化隨機數生成器
rng = np.random.default_rng(seed=42)
n = 60

# 2. 生成 60 筆學生數據
df = pd.DataFrame(
    {
        "student_id": [f"S{i:03d}" for i in range(1, n + 1)],
        "study_hours": np.round(rng.uniform(1.0, 10.0, size=n), 1),
        "sleep_hours": np.round(rng.uniform(4.0, 9.0, size=n), 1),
        "screen_time_hours": np.round(rng.uniform(1.0, 8.0, size=n), 1),
        "attendance_rate": np.round(rng.uniform(0.6, 1.0, size=n), 2),
    }
)

# 3. 自動轉換出席率單位並計算分數
attendance_pct = df["attendance_rate"] * (
    100.0 if df["attendance_rate"].max() <= 1.0 else 1.0
)
random_error = rng.normal(loc=0, scale=5, size=len(df))
calculated_score = (
    50
    + (3.0 * df["study_hours"])
    + (2.0 * df["sleep_hours"])
    - (1.2 * df["screen_time_hours"])
    + (0.15 * attendance_pct)
    + random_error
)

# 4. 轉換為 0~100 整數
df["quiz_score"] = np.floor(
    np.clip(calculated_score, 0, 100) + 0.5
).astype("Int64")

# 5. 保持與原本相同的輸出寫法（最後一行直接傳回 DataFrame，不加 print）
df[
    [
        "student_id",
        "study_hours",
        "sleep_hours",
        "screen_time_hours",
        "attendance_rate",
        "quiz_score",
    ]
]
```
